# Bloch Sphere Visualization Examples

This notebook demonstrates how to use the `plot_bloch_sphere` function to visualize quantum states on the Bloch sphere.

```python
import numpy as np
import matplotlib.pyplot as plt
from bloch_sphere import plot_bloch_sphere
```

## Basic Usage

The `plot_bloch_sphere` function takes two required parameters:
- `theta`: The polar angle from the z-axis (0 to π radians)
- `phi`: The azimuthal angle in the x-y plane (0 to 2π radians)

Let's visualize some common quantum states:

### State |0⟩ (North Pole)

For the |0⟩ state, θ = 0 and φ can be any value (typically set to 0).

```python
plot_bloch_sphere(0, 0)
plt.show()
```

### State |1⟩ (South Pole)

For the |1⟩ state, θ = π and φ can be any value (typically set to 0).

```python
plot_bloch_sphere(np.pi, 0)
plt.show()
```

### State |+⟩ (Positive X-axis)

For the |+⟩ state, θ = π/2 and φ = 0.

```python
plot_bloch_sphere(np.pi/2, 0)
plt.show()
```

### State |-⟩ (Negative X-axis)

For the |-⟩ state, θ = π/2 and φ = π.

```python
plot_bloch_sphere(np.pi/2, np.pi)
plt.show()
```

### State |+i⟩ (Positive Y-axis)

For the |+i⟩ state, θ = π/2 and φ = π/2.

```python
plot_bloch_sphere(np.pi/2, np.pi/2)
plt.show()
```

### State |-i⟩ (Negative Y-axis)

For the |-i⟩ state, θ = π/2 and φ = 3π/2.

```python
plot_bloch_sphere(np.pi/2, 3*np.pi/2)
plt.show()
```

## Custom Angles

You can specify any arbitrary angles to visualize different quantum states:

```python
# Example with custom angles
theta = np.pi/4  # 45 degrees from z-axis
phi = np.pi/3    # 60 degrees in x-y plane

plot_bloch_sphere(theta, phi)
plt.show()
```

## Additional Options

The `plot_bloch_sphere` function has additional parameters to customize the visualization:

```python
# Hide angles and vector components
plot_bloch_sphere(np.pi/3, np.pi/4, show_angles=False, show_vector_components=False)
plt.show()
```

```python
# Change figure size
plot_bloch_sphere(np.pi/3, np.pi/4, figsize=(8, 8))
plt.show()
```

## Interactive Example

You can use widgets to create an interactive Bloch sphere visualization:

```python
from ipywidgets import interact, FloatSlider

def interactive_bloch_sphere(theta=np.pi/2, phi=0):
    plot_bloch_sphere(theta, phi)
    plt.show()

# Create interactive sliders
interact(interactive_bloch_sphere, 
         theta=FloatSlider(min=0, max=np.pi, step=0.01, value=np.pi/2, description='θ:'),
         phi=FloatSlider(min=0, max=2*np.pi, step=0.01, value=0, description='φ:'))
```
